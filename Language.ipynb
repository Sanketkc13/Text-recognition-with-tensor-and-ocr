```python
import os
import json

# Define the base path for the dataset
base_dataset_path = '/content/drive/My Drive/data science class 5 to 6.30/Language/Learning/'

# Corrected file names
words_txt_path = os.path.join(base_dataset_path, 'words.txt')
words_dictionary_path = os.path.join(base_dataset_path, 'words_dictionary.json')  # Corrected name
words_alpha_txt_path = os.path.join(base_dataset_path, 'words_alpha.txt')

# Preview words.txt
if os.path.exists(words_txt_path):
    with open(words_txt_path, 'r') as file:
        words_txt_data = file.readlines()
    print("Preview of words.txt (First 5 lines):")
    print(words_txt_data[:5])  # Show first 5 lines
else:
    print(f"File {words_txt_path} not found!")

# Preview words_alpha.txt
if os.path.exists(words_alpha_txt_path):
    with open(words_alpha_txt_path, 'r') as file:
        words_alpha_txt_data = file.readlines()
    print("\nPreview of words_alpha.txt (First 5 words):")
    print(words_alpha_txt_data[:5])  # Show first 5 words
else:
    print(f"File {words_alpha_txt_path} not found!")

# Preview words_dictionary.json
if os.path.exists(words_dictionary_path):
    with open(words_dictionary_path, 'r') as json_file:
        words_dictionary = json.load(json_file)
    print("\nPreview of words_dictionary.json (First 5 entries):")
    print(list(words_dictionary.items())[:5])  # Show first 5 entries from the dictionary
else:
    print(f"File {words_dictionary_path} not found!")
```

```python
import matplotlib.pyplot as plt
import tensorflow as tf
import numpy as np
import os
import json

# Load and preprocess data
base_path = '/content/drive/My Drive/data science class 5 to 6.30/Language/Learning/'

# Load words_alpha.txt (words to be recognized)
with open(os.path.join(base_path, 'words_alpha.txt'), 'r') as file:
    words_alpha_data = [line.strip() for line in file.readlines()]

# Load words_dictionary.json (dictionary for label presence)
with open(os.path.join(base_path, 'words_dictionary.json'), 'r') as file:
    words_dictionary = json.load(file)

# Prepare labels: 1 if the word is in the dictionary, 0 otherwise
output_data = [1 if word in words_dictionary else 0 for word in words_alpha_data]

# Tokenize input words
tokenizer = tf.keras.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(words_alpha_data)
input_sequences = tokenizer.texts_to_sequences(words_alpha_data)

# Pad sequences to ensure equal length
input_padded = tf.keras.preprocessing.sequence.pad_sequences(input_sequences, padding='post')

# Convert data to NumPy arrays
input_padded = np.array(input_padded)
output_data = np.array(output_data)

# Split the dataset into training and validation sets
split_index = int(0.8 * len(input_padded))
train_input, val_input = input_padded[:split_index], input_padded[split_index:]
train_output, val_output = output_data[:split_index], output_data[split_index:]

# Define a simple neural network model for word recognition
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=len(tokenizer.word_index) + 1, output_dim=64),  # Removed input_length
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')  # Binary classification
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model and capture the history
history = model.fit(train_input, train_output, epochs=10, validation_data=(val_input, val_output), batch_size=32)

# Plot Loss Over Epochs
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

# Plot Accuracy Over Epochs
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Show the plots
plt.tight_layout()
plt.show()

# Save the model in the recommended Keras format
model.save('word_recognition_model.keras')
```

```python
import tensorflow as tf

# Define a Convolutional Neural Network (CNN) for text classification
model = tf.keras.Sequential([
    # Embedding layer that converts words into word embeddings
    tf.keras.layers.Embedding(input_dim=len(tokenizer.word_index) + 1,
                              output_dim=128,
                              input_length=train_input.shape[1]),

    # Convolutional layer with 32 filters and kernel size of 3, followed by max pooling
    tf.keras.layers.Conv1D(32, 3, activation='relu'),
    tf.keras.layers.MaxPooling1D(pool_size=2),

    # Convolutional layer with 64 filters and kernel size of 3, followed by max pooling
    tf.keras.layers.Conv1D(64, 3, activation='relu'),
    tf.keras.layers.MaxPooling1D(pool_size=2),

    # Convolutional layer with 128 filters and kernel size of 3, followed by max pooling
    tf.keras.layers.Conv1D(128, 3, activation='relu'),
    tf.keras.layers.MaxPooling1D(pool_size=2),

    # Flatten the 2D output to 1D
    tf.keras.layers.Flatten(),

    # Fully connected layer with 128 units and ReLU activation
    tf.keras.layers.Dense(128, activation='relu'),

    # Output layer with 1 unit and sigmoid activation for binary classification
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compile the model with Adam optimizer and binary crossentropy loss function
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Show model summary to check architecture
model.summary()
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

# Evaluate the model on validation/test data
try:
    loss, accuracy = model.evaluate(val_input, val_output, verbose=1)
    print(f"Validation Loss: {loss:.4f}")
    print(f"Validation Accuracy: {accuracy:.4f}")
except Exception as e:
    print(f"Error during evaluation: {e}")

# Make predictions on the validation data
try:
    predictions = model.predict(val_input)
    predicted_labels = (predictions > 0.5).astype(int).flatten()  # Convert probabilities to binary labels

    # Calculate classification metrics
    print("\nClassification Report:")
    print(classification_report(val_output, predicted_labels, target_names=['Absent', 'Present']))

    # Calculate and display confusion matrix
    conf_matrix = confusion_matrix(val_output, predicted_labels)
    print("\nConfusion Matrix:")
    print(conf_matrix)
except Exception as e:
    print(f"Error during predictions or metrics calculation: {e}")
```

```python
import tensorflow as tf
import numpy as np
import pytesseract
from PIL import Image
from google.colab import files

# Ensure Tesseract OCR is installed
!sudo apt-get update
!sudo apt-get install tesseract-ocr
!pip install pytesseract

# Set Tesseract path explicitly
pytesseract.pytesseract.tesseract_cmd = r'/usr/bin/tesseract'  # This line is crucial

# Load the trained model
model = tf.keras.models.load_model('word_recognition_model.keras')

# Load the tokenizer used during training
tokenizer = tf.keras.preprocessing.text.Tokenizer()
with open('/content/drive/My Drive/data science class 5 to 6.30/Language/Learning/words_alpha.txt', 'r') as file:
    words_alpha_data = [line.strip() for line in file.readlines()]
tokenizer.fit_on_texts(words_alpha_data)

# Function to extract words from an image using OCR
def extract_words_from_image(image_path):
    # Load the image
    image = Image.open(image_path)

    # Use Tesseract to extract text
    extracted_text = pytesseract.image_to_string(image)

    # Split into individual words
    words = extracted_text.split()
    return words

# Function to predict words using the trained model
def predict_words_from_image(image_path):
    # Extract words from the image
    words = extract_words_from_image(image_path)

    print(f"Extracted Words: {words}")

    # Tokenize and pad the words
    sequences = tokenizer.texts_to_sequences(words)

    # Check if padding is needed
    max_len = 32  # Set the max length of your input (should match the model's training)
    padded_sequences = tf.keras.preprocessing.sequence.pad_sequences(sequences, maxlen=max_len, padding='post')

    # Ensure the shape is compatible with the model's input
    padded_sequences = np.array(padded_sequences)  # Convert to numpy array if it's not already

    print(f"Padded Sequences Shape: {padded_sequences.shape}")  # Debugging output

    # Check the model input shape for compatibility
    print(f"Model input shape: {model.input_shape}")  # Debugging output

    # Ensure the correct dtype (usually float32 for neural networks)
    padded_sequences = padded_sequences.astype(np.float32)

    # Check if the batch dimension is being added automatically (e.g., shape of (None, 32))
    if len(padded_sequences.shape) == 1:
        padded_sequences = np.expand_dims(padded_sequences, axis=0)  # Add batch dimension

    # Predict using the model
    try:
        predictions = model.predict(padded_sequences, verbose=0)
        print("Prediction Successful")
    except Exception as e:
        print(f"Prediction Error: {e}")
        return

    # Convert probabilities to binary labels
    predicted_labels = (predictions > 0.5).astype(int).flatten()

    # Display predictions
    for word, probability, label in zip(words, predictions.flatten(), predicted_labels):
        print(f"Word: '{word}' - Probability: {probability:.4f}, Predicted Label: {'Present' if label == 1 else 'Absent'}")

# Upload an image
uploaded = files.upload()
image_path = next(iter(uploaded.keys()))  # Get the uploaded file name

# Predict words in the image
predict_words_from_image(image_path)
```

```python
# Install EasyOCR and OpenCV
!pip install easyocr
!apt-get install -y python3-opencv

import easyocr
import numpy as np
import cv2
from PIL import Image
from google.colab import files

# Initialize the EasyOCR reader
reader = easyocr.Reader(['en'])

# Function to extract words from an image using EasyOCR
def extract_words_from_image(image_path):
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    results = reader.readtext(image)
    words = [result[1] for result in results]
    return words

# Function to predict words using the trained model
def predict_words_from_image(image_path):
    words = extract_words_from_image(image_path)
    print(f"Extracted Words: {words}")

    sequences = tokenizer.texts_to_sequences(words)
    max_len = 32
    padded_sequences = tf.keras.preprocessing.sequence.pad_sequences(sequences, maxlen=max_len, padding='post')
    padded_sequences = np.array(padded_sequences)
    padded_sequences = padded_sequences.astype(np.float32)

    if len(padded_sequences.shape) == 1:
        padded_sequences = np.expand_dims(padded_sequences, axis=0)

    try:
        predictions = model.predict(padded_sequences, verbose=0)
        print("Prediction Successful")
    except Exception as e:
        print(f"Prediction Error: {e}")
        return

    predicted_labels = (predictions > 0.5).astype(int).flatten()

    for word, probability, label in zip(words, predictions.flatten(), predicted_labels):
        print(f"Word: '{word}' - Probability: {probability:.4f}, Predicted Label: {'Present' if label == 1 else 'Absent'}")

# Upload an image
uploaded = files.upload()
image_path = next(iter(uploaded.keys()))

# Predict words in the image
predict_words_from_image(image_path)
```
